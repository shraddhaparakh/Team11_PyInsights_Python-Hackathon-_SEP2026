Important: Don't immediately replace blanks with 0

For your patienthistory.csv, we have binary disease variables such as:

peptic_ulcer_disease
liver_disease
moderate_to_severe_chronic_kidney_disease

Suppose you find:

liver_disease = NaN

Don't immediately do:

df['liver_disease'] = df['liver_disease'].fillna(0)

because:

0 = confirmed absence of disease
NaN = information is missing/unknown

These are not necessarily the same.
